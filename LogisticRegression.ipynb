```python
import pandas as pd
import seaborn as sns
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('/content/sample_data/diabetes2.csv')
print(df)
```

```python
# Select Few Important Feature form the dataset
x = df[['Pregnancies','Glucose','Age']]
y = df['Outcome']
```

```python
# Handle missing value in X
imputer = SimpleImputer(strategy='mean')  # Initialize the imputer
x_imputed = pd.DataFrame(imputer.fit_transform(x),columns = x.columns)  # Fit and transform the data
```

```python
# To Train Model
model = LogisticRegression(max_iter=1000)
model.fit(x_imputed,y)      # Used Imputed data for training
```
